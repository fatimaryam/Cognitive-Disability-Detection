EfficientNet-B3 + Transformer + Attention Fusion

```python
# -----------------------------
# Standard libraries
# -----------------------------
import os, json, random, warnings, platform
import numpy as np
import pandas as pd
from PIL import Image

warnings.filterwarnings("ignore")

# -----------------------------
# PyTorch / TorchVision
# -----------------------------
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torch.cuda.amp import autocast, GradScaler

import torchvision
from torchvision import transforms
import torchvision.models as models

# -----------------------------
# Sklearn
# -----------------------------
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report,
    roc_curve, auc, ConfusionMatrixDisplay,
)
from sklearn.utils.class_weight import compute_class_weight

# -----------------------------
# Viz / progress
# -----------------------------
import matplotlib.pyplot as plt
from tqdm.auto import tqdm

# ==========================================================
# SEED / DEVICE
# ==========================================================

SEED = 42
random.seed(SEED); np.random.seed(SEED)
torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False


def seed_worker(worker_id):
    worker_seed = SEED + worker_id
    np.random.seed(worker_seed)
    random.seed(worker_seed)


g = torch.Generator()
g.manual_seed(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("=" * 60)
print("Device :", DEVICE)
if DEVICE.type == "cuda":
    print("GPU :", torch.cuda.get_device_name(0))
print("=" * 60)

# ==========================================================
# PATHS
# ==========================================================

DATASET_ROOT = "/content/drive/MyDrive/MCI_dataset"
TRAIN_CSV = os.path.join(DATASET_ROOT, "split/train.csv")
VAL_CSV   = os.path.join(DATASET_ROOT, "split/val.csv")
TEST_CSV  = os.path.join(DATASET_ROOT, "split/test.csv")

# ==========================================================
# MODEL / TRAINING CONFIG — matched to your EfficientNet-B3 baseline
# ==========================================================

BACKBONE = "efficientnet_b3"
FUSION = "transformer_full_feature_concat_experiment_B"

OUTPUT_DIR = f"/content/drive/MyDrive/MCI_Results_{BACKBONE}_TransformerConcatFusion"
os.makedirs(OUTPUT_DIR, exist_ok=True)
CHECKPOINT_PATH = os.path.join(OUTPUT_DIR, "best_model.pth")

NUM_CLASSES = 2
CLASS_NAMES = ["Normal", "MCI"]
IMAGE_SIZE = 300               # EfficientNet-B3 native — MATCHED to baseline
BATCH_SIZE = 16
NUM_WORKERS = 2
EPOCHS = 50
LEARNING_RATE = 1e-4           # MATCHED to baseline — unchanged this round
WEIGHT_DECAY = 1e-4            # MATCHED to baseline — unchanged this round
UNFREEZE_FRACTION = 0.30       # MATCHED to baseline — unchanged this round
GRAD_CLIP_NORM = 1.0
EARLY_STOPPING_PATIENCE = 10
USE_AMP = torch.cuda.is_available()

# Transformer config
EMBED_DIM = 768                # EfficientNet stays 1536; transformer works in 768
NUM_HEADS = 8
NUM_TRANSFORMER_LAYERS = 1     # kept minimal — 3 tokens don't need depth
TRANSFORMER_FFN_DIM = 1536
TRANSFORMER_DROPOUT = 0.10     # reduced from 0.20 — small dataset, single layer
ATTENTION_HIDDEN = 256

print("\nConfiguration")
print("-" * 60)
print(f"Backbone            : {BACKBONE}")
print(f"Fusion              : {FUSION}")
print(f"Image Size          : {IMAGE_SIZE}")
print(f"Batch Size          : {BATCH_SIZE}")
print(f"Epochs              : {EPOCHS}")
print(f"Learning Rate       : {LEARNING_RATE}")
print(f"Weight Decay        : {WEIGHT_DECAY}")
print(f"Unfreeze Frac.      : {UNFREEZE_FRACTION}")
print(f"Embed Dim           : {EMBED_DIM}")
print(f"Transformer Dropout : {TRANSFORMER_DROPOUT}")
print(f"Mixed Precision     : {USE_AMP}")
print("-" * 60)


def save_run_manifest():
    manifest = {
        "backbone": BACKBONE, "fusion": FUSION, "seed": SEED,
        "image_size": IMAGE_SIZE, "batch_size": BATCH_SIZE, "epochs": EPOCHS,
        "unfreeze_fraction": UNFREEZE_FRACTION,
        "embed_dim": EMBED_DIM, "num_heads": NUM_HEADS,
        "num_transformer_layers": NUM_TRANSFORMER_LAYERS,
        "transformer_ffn_dim": TRANSFORMER_FFN_DIM,
        "transformer_dropout": TRANSFORMER_DROPOUT,
        "attention_hidden": ATTENTION_HIDDEN,
        "learning_rate": LEARNING_RATE, "weight_decay": WEIGHT_DECAY,
        "grad_clip_norm": GRAD_CLIP_NORM,
        "torch_version": torch.__version__,
        "torchvision_version": torchvision.__version__,
        "python_version": platform.python_version(),
        "cuda_available": torch.cuda.is_available(),
        "gpu_name": torch.cuda.get_device_name(0) if torch.cuda.is_available() else None,
    }
    with open(os.path.join(OUTPUT_DIR, "run_manifest.json"), "w") as f:
        json.dump(manifest, f, indent=2)
    print("Run manifest saved to", os.path.join(OUTPUT_DIR, "run_manifest.json"))


save_run_manifest()

# ==========================================================
# DATASET
# ==========================================================

class MCIDataset(Dataset):
    def __init__(self, csv_file, transform=None):
        self.df = pd.read_csv(csv_file, dtype={"patient_id": str}).reset_index(drop=True)
        self.transform = transform
        print(f"{os.path.basename(csv_file)} : {len(self.df)} patients")

    def __len__(self):
        return len(self.df)

    def load_image(self, image_path):
        image = Image.open(image_path).convert("RGB")
        if self.transform is not None:
            image = self.transform(image)
        return image

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        return {
            "patient_id": row["patient_id"],
            "clock": self.load_image(row["clock_path"]),
            "copy": self.load_image(row["copy_path"]),
            "trail": self.load_image(row["trail_path"]),
            "label": torch.tensor(int(row["label"]), dtype=torch.long),
            "moca": float(row["MoCA_score"]),
        }


train_df = pd.read_csv(TRAIN_CSV)
val_df = pd.read_csv(VAL_CSV)
test_df = pd.read_csv(TEST_CSV)

print()
print("Train :", len(train_df))
print("Validation :", len(val_df))
print("Test :", len(test_df))

# ==========================================================
# TRANSFORMS — no RandomHorizontalFlip: clock numerals and
# trail direction are semantically meaningful.
# ==========================================================

IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomRotation(10),
    transforms.RandomAffine(degrees=0, translate=(0.05, 0.05), scale=(0.9, 1.1)),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])

val_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])

train_dataset = MCIDataset(TRAIN_CSV, transform=train_transform)
val_dataset   = MCIDataset(VAL_CSV,   transform=val_transform)
test_dataset  = MCIDataset(TEST_CSV,  transform=val_transform)

# ==========================================================
# DATALOADERS
# ==========================================================

PIN_MEMORY = True

train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE, shuffle=True,
    num_workers=NUM_WORKERS, pin_memory=PIN_MEMORY,
    persistent_workers=(NUM_WORKERS > 0), drop_last=True,
    worker_init_fn=seed_worker, generator=g,
)
val_loader = DataLoader(
    val_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=PIN_MEMORY,
    persistent_workers=(NUM_WORKERS > 0),
)
test_loader = DataLoader(
    test_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=PIN_MEMORY,
    persistent_workers=(NUM_WORKERS > 0),
)

print(f"Train : {len(train_dataset)} patients")
print(f"Validation : {len(val_dataset)} patients")
print(f"Test : {len(test_dataset)} patients")

class_weights = compute_class_weight(
    class_weight="balanced", classes=np.unique(train_df["label"]), y=train_df["label"]
)
class_weights = torch.tensor(class_weights, dtype=torch.float32, device=DEVICE)
print("\nClass Weights"); print(class_weights)

# ==========================================================
# BACKBONE — plain EfficientNet-B3, NO CBAM this round
# (isolate the transformer's effect first)
# ==========================================================

def build_backbone(name):
    if name.lower() == "efficientnet_b3":
        weights = models.EfficientNet_B3_Weights.IMAGENET1K_V1
        backbone = models.efficientnet_b3(weights=weights)
        feature_dim = backbone.classifier[1].in_features   # 1536
        backbone.classifier = nn.Identity()
    elif name.lower() == "densenet121":
        weights = models.DenseNet121_Weights.IMAGENET1K_V1
        backbone = models.densenet121(weights=weights)
        feature_dim = backbone.classifier.in_features
        backbone.classifier = nn.Identity()
    else:
        raise ValueError("Unsupported backbone")
    return backbone, feature_dim


# ==========================================================
# ATTENTION POOL — one extra summary of the transformer output,
# kept as an ADDITIONAL fused component (not a replacement).
# ==========================================================

class AttentionPool(nn.Module):
    def __init__(self, embed_dim, hidden_dim=256):
        super().__init__()
        self.attention = nn.Sequential(
            nn.Linear(embed_dim, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, 1),
        )

    def forward(self, x):
        # x: [B, 3, D]
        scores = self.attention(x).squeeze(-1)
        weights = torch.softmax(scores, dim=1)
        pooled = torch.sum(x * weights.unsqueeze(-1), dim=1)
        return pooled, weights


# ==========================================================
# FULL-FEATURE-PRESERVING FUSION
# Concatenates: original EfficientNet features (3*1536=4608) +
# full transformer output (3*768=2304) + attention-pooled
# transformer summary (768) = 7680-d fused vector. No gate, no
# information discarded — the classifier decides what to weight.
# ==========================================================

class ResidualAttentionFusion(nn.Module):

    def __init__(self, embed_dim, attention_hidden=256):
        super().__init__()
        self.attention_pool = AttentionPool(embed_dim=embed_dim, hidden_dim=attention_hidden)

    def forward(self, transformer_features, original_features):
        # transformer_features: [B, 3, embed_dim]   (e.g. 3 x 768)
        # original_features:    [B, 3, feature_dim]  (e.g. 3 x 1536, RAW backbone output)
        transformer_pooled, attention_weights = self.attention_pool(transformer_features)

        transformer_concat = transformer_features.flatten(1)   # [B, 3*embed_dim]
        original_concat = original_features.flatten(1)         # [B, 3*feature_dim]

        fused = torch.cat([original_concat, transformer_concat, transformer_pooled], dim=1)
        return fused, attention_weights


# ==========================================================
# CORRECTED FOCAL LOSS
# Same mathematical result as before, implemented via the more
# standard log_softmax / gather / nll_loss path.
# ==========================================================

class FocalLoss(nn.Module):
    def __init__(self, alpha=None, gamma=2.0):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma

    def forward(self, inputs, targets):
        log_probs = F.log_softmax(inputs, dim=1)
        log_pt = log_probs.gather(1, targets.unsqueeze(1)).squeeze(1)
        pt = log_pt.exp()
        ce = F.nll_loss(log_probs, targets, weight=self.alpha, reduction="none")
        loss = ((1.0 - pt) ** self.gamma) * ce
        return loss.mean()


# ==========================================================
# MAIN MODEL
# EfficientNet-B3 -> project to 768 -> Transformer (cross-modal
# attention across clock/copy/trail) -> concat[original 4608,
# transformer 2304, attention-pooled 768] = 7680 -> classifier
# ==========================================================

class MCITransformerConcatFusion(nn.Module):

    def __init__(self, backbone_name="efficientnet_b3", num_classes=2,
                 embed_dim=768, num_heads=8, num_layers=1, ffn_dim=1536,
                 transformer_dropout=0.10, attention_hidden=256,
                 unfreeze_fraction=0.30):
        super().__init__()

        self.backbone_name = backbone_name.lower()
        self.backbone, feature_dim = build_backbone(backbone_name)   # feature_dim = 1536

        for param in self.backbone.parameters():
            param.requires_grad = False

        feature_children = list(self.backbone.features.children())
        n = len(feature_children)
        start = int(n * (1 - unfreeze_fraction))
        for child in feature_children[start:]:
            for param in child.parameters():
                param.requires_grad = True

        self.project = nn.Sequential(
            nn.Linear(feature_dim, embed_dim),
            nn.GELU(),
            nn.LayerNorm(embed_dim),
            nn.Dropout(0.10),
        )

        self.modality_embedding = nn.Parameter(torch.zeros(1, 3, embed_dim))
        nn.init.normal_(self.modality_embedding, mean=0.0, std=0.02)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=embed_dim, nhead=num_heads, dim_feedforward=ffn_dim,
            dropout=transformer_dropout, activation="gelu",
            batch_first=True, norm_first=True,
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        self.fusion = ResidualAttentionFusion(embed_dim=embed_dim, attention_hidden=attention_hidden)

        fused_dim = (feature_dim * 3) + (embed_dim * 3) + embed_dim   # 4608 + 2304 + 768 = 7680

        self.classifier = nn.Sequential(
            nn.LayerNorm(fused_dim), nn.Dropout(0.40),
            nn.Linear(fused_dim, 1024), nn.GELU(),
            nn.Dropout(0.30),
            nn.Linear(1024, 256), nn.GELU(),
            nn.Dropout(0.20),
            nn.Linear(256, num_classes),
        )

    def forward(self, clock, copy, trail, return_attention=False):
        f_clock = self.backbone(clock)   # [B, 1536] — RAW backbone features
        f_copy = self.backbone(copy)
        f_trail = self.backbone(trail)

        original_stack = torch.stack([f_clock, f_copy, f_trail], dim=1)  # [B, 3, 1536]

        x = self.project(original_stack)               # [B, 3, embed_dim]
        x_with_modality = x + self.modality_embedding

        transformer_out = self.transformer(x_with_modality)  # [B, 3, embed_dim]

        # Fuse RAW original features (not the projected `x`) with the
        # transformer output — this preserves everything the baseline
        # concat model saw, plus what the transformer learned.
        fused, attention_weights = self.fusion(transformer_out, original_stack)

        logits = self.classifier(fused)

        if return_attention:
            return logits, attention_weights
        return logits


model = MCITransformerConcatFusion(
    backbone_name=BACKBONE, num_classes=NUM_CLASSES, embed_dim=EMBED_DIM,
    num_heads=NUM_HEADS, num_layers=NUM_TRANSFORMER_LAYERS,
    ffn_dim=TRANSFORMER_FFN_DIM, transformer_dropout=TRANSFORMER_DROPOUT,
    attention_hidden=ATTENTION_HIDDEN, unfreeze_fraction=UNFREEZE_FRACTION,
).to(DEVICE)

trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
total = sum(p.numel() for p in model.parameters())
print("=" * 60)
print("Model Summary")
print("=" * 60)
print("Backbone :", BACKBONE, "| Fusion :", FUSION)
print(f"Trainable Parameters : {trainable:,}")
print(f"Total Parameters     : {total:,}")
print("=" * 60)

# ==========================================================
# LOSS / OPTIMIZER / SCHEDULER
# ==========================================================

criterion = FocalLoss(alpha=class_weights, gamma=2.0)

trainable_params = [p for p in model.parameters() if p.requires_grad]
optimizer = optim.AdamW(trainable_params, lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)

scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="min", factor=0.5, patience=3)
scaler = GradScaler(enabled=USE_AMP)

history = {"train_loss": [], "train_acc": [], "val_loss": [], "val_acc": [], "val_auc": [], "lr": []}

# ==========================================================
# TRAIN / VALIDATE
# ==========================================================

def train_one_epoch(model, loader):
    model.train()
    running_loss, running_correct, total = 0.0, 0, 0
    loop = tqdm(loader, leave=False)
    for batch in loop:
        clock = batch["clock"].to(DEVICE, non_blocking=True)
        copy = batch["copy"].to(DEVICE, non_blocking=True)
        trail = batch["trail"].to(DEVICE, non_blocking=True)
        labels = batch["label"].to(DEVICE)

        optimizer.zero_grad()
        with autocast(enabled=USE_AMP):
            outputs = model(clock, copy, trail)
            loss = criterion(outputs, labels)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), GRAD_CLIP_NORM)
        scaler.step(optimizer)
        scaler.update()

        preds = outputs.argmax(dim=1)
        running_loss += loss.item() * labels.size(0)
        running_correct += (preds == labels).sum().item()
        total += labels.size(0)
        loop.set_postfix(loss=f"{loss.item():.4f}", acc=f"{100*running_correct/total:.2f}%")

    return running_loss / total, running_correct / total


def validate(model, loader):
    model.eval()
    running_loss, running_correct, total = 0.0, 0, 0
    y_true, y_prob = [], []
    with torch.no_grad():
        for batch in tqdm(loader, leave=False):
            clock = batch["clock"].to(DEVICE, non_blocking=True)
            copy = batch["copy"].to(DEVICE, non_blocking=True)
            trail = batch["trail"].to(DEVICE, non_blocking=True)
            labels = batch["label"].to(DEVICE)

            with autocast(enabled=USE_AMP):
                outputs = model(clock, copy, trail)
                loss = criterion(outputs, labels)

            probs = torch.softmax(outputs, dim=1)
            preds = outputs.argmax(dim=1)
            running_loss += loss.item() * labels.size(0)
            running_correct += (preds == labels).sum().item()
            total += labels.size(0)
            y_true.extend(labels.cpu().numpy())
            y_prob.extend(probs[:, 1].cpu().numpy())

    try:
        epoch_auc = roc_auc_score(y_true, y_prob)
    except Exception:
        epoch_auc = 0.5

    return running_loss / total, running_correct / total, epoch_auc


# ==========================================================
# TRAINING LOOP
# Checkpoint selection AND early stopping BOTH track val_auc now,
# so the two objectives agree (previously loss-based early stopping
# could fire while an AUC-based checkpoint was still improving).
# ==========================================================

print("=" * 70); print("Training Started"); print("=" * 70)

best_auc = -np.inf
patience_counter = 0

for epoch in range(EPOCHS):
    train_loss, train_acc = train_one_epoch(model, train_loader)
    val_loss, val_acc, val_auc = validate(model, val_loader)
    scheduler.step(val_loss)

    history["train_loss"].append(train_loss)
    history["train_acc"].append(train_acc)
    history["val_loss"].append(val_loss)
    history["val_acc"].append(val_acc)
    history["val_auc"].append(val_auc)
    history["lr"].append(optimizer.param_groups[0]["lr"])

    print(
        f"Epoch {epoch+1:02d}/{EPOCHS}"
        f" | Train Loss {train_loss:.4f}"
        f" | Train Acc {train_acc:.4f}"
        f" | Val Loss {val_loss:.4f}"
        f" | Val Acc {val_acc:.4f}"
        f" | Val AUC {val_auc:.4f}"
        f" | LR {optimizer.param_groups[0]['lr']:.2e}"
    )

    if val_auc > best_auc + 1e-4:
        best_auc = val_auc
        patience_counter = 0
        torch.save(model.state_dict(), CHECKPOINT_PATH)
        print("★★★★★ Best AUC Model Updated ★★★★★")
    else:
        patience_counter += 1
        print(f"EarlyStopping: {patience_counter}/{EARLY_STOPPING_PATIENCE}")

    if patience_counter >= EARLY_STOPPING_PATIENCE:
        print("\nEarly stopping activated.")
        break

print("\n" + "=" * 70)
print("Training Finished")
print("=" * 70)
print(f"Best Validation AUC : {best_auc:.4f}")

# ==========================================================
# TEST EVALUATION (threshold tuned on val)
# ==========================================================

print("=" * 60); print("Loading Best Model"); print("=" * 60)
model.load_state_dict(torch.load(CHECKPOINT_PATH))
model.eval()

val_probs, val_labels = [], []
with torch.no_grad():
    for batch in tqdm(val_loader):
        clock = batch["clock"].to(DEVICE)
        copy = batch["copy"].to(DEVICE)
        trail = batch["trail"].to(DEVICE)
        labels = batch["label"].cpu().numpy()
        outputs = model(clock, copy, trail)
        probs = torch.softmax(outputs, dim=1)[:, 1]
        val_probs.extend(probs.cpu().numpy())
        val_labels.extend(labels)

best_threshold, best_f1 = 0.50, 0
for t in np.arange(0.20, 0.81, 0.01):
    preds = (np.array(val_probs) >= t).astype(int)
    f1 = f1_score(val_labels, preds)
    if f1 > best_f1:
        best_f1, best_threshold = f1, t

print(f"\nBest Threshold = {best_threshold:.2f}")
print(f"Validation F1 = {best_f1:.4f}")

test_probs, test_preds, test_labels, test_attn, test_patient_ids = [], [], [], [], []
running_loss, total = 0, 0
with torch.no_grad():
    for batch in tqdm(test_loader):
        clock = batch["clock"].to(DEVICE)
        copy = batch["copy"].to(DEVICE)
        trail = batch["trail"].to(DEVICE)
        labels = batch["label"].to(DEVICE)

        outputs, attn_w = model(clock, copy, trail, return_attention=True)
        loss = criterion(outputs, labels)
        probs = torch.softmax(outputs, dim=1)[:, 1]
        preds = (probs >= best_threshold).long()

        running_loss += loss.item() * labels.size(0)
        total += labels.size(0)
        test_probs.extend(probs.cpu().numpy())
        test_preds.extend(preds.cpu().numpy())
        test_labels.extend(labels.cpu().numpy())
        test_attn.extend(attn_w.cpu().numpy())
        test_patient_ids.extend(batch["patient_id"])

test_loss = running_loss / total
accuracy = accuracy_score(test_labels, test_preds)
precision = precision_score(test_labels, test_preds)
recall = recall_score(test_labels, test_preds)
f1 = f1_score(test_labels, test_preds)
test_auc = roc_auc_score(test_labels, test_probs)
cm = confusion_matrix(test_labels, test_preds)
tn, fp, fn, tp = cm.ravel()
specificity = tn / (tn + fp)

print("\n" + "=" * 70)
print("TEST RESULTS")
print("=" * 70)
print(f"Threshold   : {best_threshold:.2f}")
print(f"Test Loss   : {test_loss:.4f}")
print(f"Accuracy    : {accuracy:.4f}")
print(f"Precision   : {precision:.4f}")
print(f"Recall      : {recall:.4f}")
print(f"Specificity : {specificity:.4f}")
print(f"F1 Score    : {f1:.4f}")
print(f"ROC AUC     : {test_auc:.4f}")
print("=" * 70)
print("\nConfusion Matrix"); print(cm)
print("\nClassification Report\n")
print(classification_report(test_labels, test_preds, target_names=CLASS_NAMES))

metrics = pd.DataFrame({
    "Accuracy": [accuracy], "Precision": [precision], "Recall": [recall],
    "Specificity": [specificity], "F1": [f1], "AUC": [test_auc],
    "Threshold": [best_threshold],
})
metrics_path = os.path.join(OUTPUT_DIR, "test_metrics.csv")
metrics.to_csv(metrics_path, index=False)
print("\nMetrics saved to"); print(metrics_path)

# ==========================================================
# *** NEW *** CLASS-WISE METRICS EXPORT
# Per-class precision / recall / f1 / support, plus accuracy and
# macro/weighted averages, saved as a CSV and plotted as a bar chart.
# ==========================================================

report_dict = classification_report(
    test_labels, test_preds, target_names=CLASS_NAMES, output_dict=True
)
classwise_df = pd.DataFrame(report_dict).transpose()
classwise_path = os.path.join(OUTPUT_DIR, "classwise_metrics.csv")
classwise_df.to_csv(classwise_path)
print("\nClass-wise metrics:")
print(classwise_df)
print("\nClass-wise metrics saved to", classwise_path)

plt.figure(figsize=(7, 5))
metric_names = ["precision", "recall", "f1-score"]
x = np.arange(len(CLASS_NAMES))
width = 0.25
for i, m in enumerate(metric_names):
    vals = [report_dict[c][m] for c in CLASS_NAMES]
    bars = plt.bar(x + i * width, vals, width, label=m)
    for b, v in zip(bars, vals):
        plt.text(b.get_x() + b.get_width() / 2, v + 0.01, f"{v:.2f}",
                  ha="center", va="bottom", fontsize=8)
plt.xticks(x + width, CLASS_NAMES)
plt.ylim(0, 1.05)
plt.ylabel("Score")
plt.title(f"Class-wise Precision / Recall / F1 — {BACKBONE} (Exp. B)")
plt.legend()
plt.grid(True, axis="y", alpha=0.3)
classwise_bar_path = os.path.join(OUTPUT_DIR, "classwise_metrics_bar.png")
plt.savefig(classwise_bar_path, dpi=300, bbox_inches="tight")
plt.show()

# ==========================================================
# *** NEW *** PER-CLASS ROC CURVES
# One-vs-rest: each class in turn is treated as the "positive"
# class. For a binary problem this gives the MCI-positive curve
# (same as the main ROC AUC) and its mirror for Normal-positive,
# so both classes get their own labeled curve/AUC on one plot.
# ==========================================================

test_labels_arr = np.array(test_labels)
test_probs_arr = np.array(test_probs)      # P(class = MCI)

plt.figure(figsize=(6, 6))

# MCI (class 1) as positive
fpr_mci, tpr_mci, _ = roc_curve(test_labels_arr, test_probs_arr)
auc_mci = auc(fpr_mci, tpr_mci)
plt.plot(fpr_mci, tpr_mci, lw=2, label=f"MCI (AUC = {auc_mci:.3f})")

# Normal (class 0) as positive
normal_true = 1 - test_labels_arr
normal_probs = 1 - test_probs_arr
fpr_normal, tpr_normal, _ = roc_curve(normal_true, normal_probs)
auc_normal = auc(fpr_normal, tpr_normal)
plt.plot(fpr_normal, tpr_normal, lw=2, label=f"Normal (AUC = {auc_normal:.3f})")

plt.plot([0, 1], [0, 1], linestyle="--", linewidth=1, color="gray")
plt.xlabel("False Positive Rate", fontsize=12)
plt.ylabel("True Positive Rate", fontsize=12)
plt.title(f"Per-Class ROC Curves — {BACKBONE} + Transformer (Exp. B)", fontsize=14)
plt.legend(loc="lower right")
plt.grid(True)
classwise_roc_path = os.path.join(OUTPUT_DIR, "classwise_roc_curve.png")
plt.savefig(classwise_roc_path, dpi=300, bbox_inches="tight")
plt.show()

# ==========================================================
# 95% BOOTSTRAP CONFIDENCE INTERVAL FOR TEST ACCURACY
# n=138 test patients — a couple points of accuracy can be noise.
# ==========================================================

def bootstrap_accuracy_ci(y_true, y_pred, n_boot=2000, seed=SEED):
    rng = np.random.RandomState(seed)
    y_true = np.array(y_true); y_pred = np.array(y_pred)
    n = len(y_true)
    boot_accs = []
    for _ in range(n_boot):
        idx = rng.randint(0, n, n)
        boot_accs.append(accuracy_score(y_true[idx], y_pred[idx]))
    lower, upper = np.percentile(boot_accs, [2.5, 97.5])
    return lower, upper


ci_lower, ci_upper = bootstrap_accuracy_ci(test_labels, test_preds)
print(f"\nTest Accuracy 95% CI (bootstrap, n=2000): [{ci_lower:.4f}, {ci_upper:.4f}]")

# ==========================================================
# ATTENTION-POOL INTERPRETABILITY EXPORT
# ==========================================================

test_attn = np.array(test_attn)
attn_df = pd.DataFrame({
    "patient_id": test_patient_ids,
    "attn_clock": test_attn[:, 0],
    "attn_copy": test_attn[:, 1],
    "attn_trail": test_attn[:, 2],
})
attn_path = os.path.join(OUTPUT_DIR, "attention_weights_test.csv")
attn_df.to_csv(attn_path, index=False)
print("\nMean attention weight by drawing (test set, transformer branch):")
print(f"  clock : {test_attn[:, 0].mean():.4f}")
print(f"  copy  : {test_attn[:, 1].mean():.4f}")
print(f"  trail : {test_attn[:, 2].mean():.4f}")

plt.figure(figsize=(5, 4))
plt.bar(["Clock", "Copy", "Trail"], test_attn.mean(axis=0))
plt.ylabel("Mean attention weight (test set)")
plt.title(f"Per-drawing attention weight — {BACKBONE} Transformer (Exp. B)")
attn_bar_path = os.path.join(OUTPUT_DIR, "attention_weights_bar.png")
plt.savefig(attn_bar_path, dpi=300, bbox_inches="tight")
plt.show()

# ==========================================================
# ROC CURVE, CONFUSION MATRIX, PREDICTIONS
# ==========================================================

fpr, tpr, _ = roc_curve(test_labels, test_probs)
roc_auc_val = auc(fpr, tpr)

plt.figure(figsize=(6, 6))
plt.plot(fpr, tpr, lw=2, label=f"AUC = {roc_auc_val:.3f}")
plt.plot([0, 1], [0, 1], linestyle="--", linewidth=1)
plt.xlabel("False Positive Rate", fontsize=12)
plt.ylabel("True Positive Rate", fontsize=12)
plt.title(f"ROC Curve — {BACKBONE} + Transformer (Exp. B)", fontsize=14)
plt.legend(loc="lower right")
plt.grid(True)
roc_path = os.path.join(OUTPUT_DIR, "roc_curve.png")
plt.savefig(roc_path, dpi=300, bbox_inches="tight")
plt.show()

disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=CLASS_NAMES)
fig, ax = plt.subplots(figsize=(6, 6))
disp.plot(cmap="Blues", ax=ax, colorbar=False, values_format="d")
plt.title("Confusion Matrix", fontsize=14)
cm_path = os.path.join(OUTPUT_DIR, "confusion_matrix.png")
plt.savefig(cm_path, dpi=300, bbox_inches="tight")
plt.show()

# ==========================================================
# *** NEW *** NORMALIZED (ROW-WISE) CONFUSION MATRIX
# Same counts as confusion_matrix.png, but each row (true class)
# is divided by its total, so you get the per-class recall as a
# percentage — much easier to read with imbalanced classes
# (98 Normal vs 40 MCI in this test set).
# ==========================================================

cm_norm = cm.astype("float") / cm.sum(axis=1, keepdims=True)
disp_norm = ConfusionMatrixDisplay(confusion_matrix=cm_norm, display_labels=CLASS_NAMES)
fig, ax = plt.subplots(figsize=(6, 6))
disp_norm.plot(cmap="Blues", ax=ax, colorbar=False, values_format=".2f")
plt.title("Normalized Confusion Matrix (row = true label)", fontsize=14)
cm_norm_path = os.path.join(OUTPUT_DIR, "confusion_matrix_normalized.png")
plt.savefig(cm_norm_path, dpi=300, bbox_inches="tight")
plt.show()

predictions = pd.DataFrame({
    "True_Label": test_labels, "Predicted_Label": test_preds, "Probability_MCI": test_probs
})
prediction_path = os.path.join(OUTPUT_DIR, "test_predictions.csv")
predictions.to_csv(prediction_path, index=False)

plt.figure(figsize=(8, 5))
plt.plot(history["train_loss"], label="Train Loss", linewidth=2)
plt.plot(history["val_loss"], label="Validation Loss", linewidth=2)
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title(f"Loss — {BACKBONE} + Transformer (Exp. B)")
plt.legend(); plt.grid(True)
loss_curve_path = os.path.join(OUTPUT_DIR, "loss_curve.png")
plt.savefig(loss_curve_path, dpi=300, bbox_inches="tight")
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(history["train_acc"], label="Train Acc", linewidth=2)
plt.plot(history["val_acc"], label="Validation Acc", linewidth=2)
plt.plot(history["val_auc"], label="Validation AUC", linewidth=2, linestyle="--")
plt.xlabel("Epoch"); plt.ylabel("Score"); plt.title(f"Acc/AUC — {BACKBONE} + Transformer (Exp. B)")
plt.legend(); plt.grid(True)
acc_auc_path = os.path.join(OUTPUT_DIR, "acc_auc_curve.png")
plt.savefig(acc_auc_path, dpi=300, bbox_inches="tight")
plt.show()

print("=" * 60)
print("Results Saved")
print("=" * 60)
print("ROC Curve                       :", roc_path)
print("Confusion Matrix                :", cm_path)
print("Confusion Matrix (normalized)   :", cm_norm_path)
print("Class-wise Metrics CSV          :", classwise_path)
print("Class-wise Metrics Bar Chart    :", classwise_bar_path)
print("Class-wise ROC Curves           :", classwise_roc_path)
print("Predictions CSV                 :", prediction_path)
print("Attention Weights CSV           :", attn_path)
print("Attention Bar Chart             :", attn_bar_path)
print("Loss Curve                      :", loss_curve_path)
print("Acc/AUC Curve                   :", acc_auc_path)
print("Metrics CSV                     :", metrics_path)
print("\nThis is Experiment B: EfficientNet-B3 baseline (~76%) + Transformer,")
print("with ALL original features preserved and concatenated (no gate, no")
print("discarded information). Compare test_metrics.csv here against your")
print("baseline directly. If B beats A, move to Experiment C/D (add CBAM")
print("back) one change at a time. If B does not beat A, that's a real,")
print("useful finding: cross-modal attention isn't adding signal on this")
print("dataset — check attention_weights_test.csv and loss_curve.png before")
print("concluding why, rather than adding more modules.")
```

```python

import os, random
import numpy as np
import pandas as pd
from PIL import Image

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

import torchvision
from torchvision import transforms
import torchvision.models as models

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, roc_curve, confusion_matrix,
)
from sklearn.utils.class_weight import compute_class_weight
from tqdm.auto import tqdm
import matplotlib.pyplot as plt

# ==========================================================
# CONFIG — MUST MATCH your original training run exactly
# ==========================================================

SEED = 42
random.seed(SEED); np.random.seed(SEED)
torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

DATASET_ROOT = "/content/drive/MyDrive/MCI_dataset"
TEST_CSV  = os.path.join(DATASET_ROOT, "split/test.csv")
VAL_CSV   = os.path.join(DATASET_ROOT, "split/val.csv")

BACKBONE = "efficientnet_b3"
OUTPUT_DIR = f"/content/drive/MyDrive/MCI_Results_{BACKBONE}_TransformerConcatFusion"
CHECKPOINT_PATH = os.path.join(OUTPUT_DIR, "best_model.pth")

NUM_CLASSES = 2
CLASS_NAMES = ["Normal", "MCI"]
IMAGE_SIZE = 300
BATCH_SIZE = 16
NUM_WORKERS = 2

EMBED_DIM = 768
NUM_HEADS = 8
NUM_TRANSFORMER_LAYERS = 1
TRANSFORMER_FFN_DIM = 1536
TRANSFORMER_DROPOUT = 0.10
ATTENTION_HIDDEN = 256
UNFREEZE_FRACTION = 0.30

IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]

# ==========================================================
# DATASET
# ==========================================================

class MCIDataset(Dataset):
    def __init__(self, csv_file, transform=None):
        self.df = pd.read_csv(csv_file, dtype={"patient_id": str}).reset_index(drop=True)
        self.transform = transform
        print(f"{os.path.basename(csv_file)} : {len(self.df)} patients")

    def __len__(self):
        return len(self.df)

    def load_image(self, image_path):
        image = Image.open(image_path).convert("RGB")
        if self.transform is not None:
            image = self.transform(image)
        return image

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        return {
            "patient_id": row["patient_id"],
            "clock": self.load_image(row["clock_path"]),
            "copy": self.load_image(row["copy_path"]),
            "trail": self.load_image(row["trail_path"]),
            "label": torch.tensor(int(row["label"]), dtype=torch.long),
        }

val_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])

val_dataset = MCIDataset(VAL_CSV, transform=val_transform)
test_dataset = MCIDataset(TEST_CSV, transform=val_transform)

val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False,
                         num_workers=NUM_WORKERS, pin_memory=True)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False,
                          num_workers=NUM_WORKERS, pin_memory=True)

# ==========================================================
# MODEL DEFINITION — must be IDENTICAL to training script
# ==========================================================

def build_backbone(name):
    if name.lower() == "efficientnet_b3":
        weights = models.EfficientNet_B3_Weights.IMAGENET1K_V1
        backbone = models.efficientnet_b3(weights=weights)
        feature_dim = backbone.classifier[1].in_features
        backbone.classifier = nn.Identity()
    else:
        raise ValueError("Unsupported backbone")
    return backbone, feature_dim


class AttentionPool(nn.Module):
    def __init__(self, embed_dim, hidden_dim=256):
        super().__init__()
        self.attention = nn.Sequential(
            nn.Linear(embed_dim, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, 1),
        )

    def forward(self, x):
        scores = self.attention(x).squeeze(-1)
        weights = torch.softmax(scores, dim=1)
        pooled = torch.sum(x * weights.unsqueeze(-1), dim=1)
        return pooled, weights


class ResidualAttentionFusion(nn.Module):
    def __init__(self, embed_dim, attention_hidden=256):
        super().__init__()
        self.attention_pool = AttentionPool(embed_dim=embed_dim, hidden_dim=attention_hidden)

    def forward(self, transformer_features, original_features):
        transformer_pooled, attention_weights = self.attention_pool(transformer_features)
        transformer_concat = transformer_features.flatten(1)
        original_concat = original_features.flatten(1)
        fused = torch.cat([original_concat, transformer_concat, transformer_pooled], dim=1)
        return fused, attention_weights


class MCITransformerConcatFusion(nn.Module):
    def __init__(self, backbone_name="efficientnet_b3", num_classes=2,
                 embed_dim=768, num_heads=8, num_layers=1, ffn_dim=1536,
                 transformer_dropout=0.10, attention_hidden=256,
                 unfreeze_fraction=0.30):
        super().__init__()
        self.backbone_name = backbone_name.lower()
        self.backbone, feature_dim = build_backbone(backbone_name)

        for param in self.backbone.parameters():
            param.requires_grad = False

        feature_children = list(self.backbone.features.children())
        n = len(feature_children)
        start = int(n * (1 - unfreeze_fraction))
        for child in feature_children[start:]:
            for param in child.parameters():
                param.requires_grad = True

        self.project = nn.Sequential(
            nn.Linear(feature_dim, embed_dim),
            nn.GELU(),
            nn.LayerNorm(embed_dim),
            nn.Dropout(0.10),
        )

        self.modality_embedding = nn.Parameter(torch.zeros(1, 3, embed_dim))
        nn.init.normal_(self.modality_embedding, mean=0.0, std=0.02)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=embed_dim, nhead=num_heads, dim_feedforward=ffn_dim,
            dropout=transformer_dropout, activation="gelu",
            batch_first=True, norm_first=True,
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
        self.fusion = ResidualAttentionFusion(embed_dim=embed_dim, attention_hidden=attention_hidden)

        fused_dim = (feature_dim * 3) + (embed_dim * 3) + embed_dim

        self.classifier = nn.Sequential(
            nn.LayerNorm(fused_dim), nn.Dropout(0.40),
            nn.Linear(fused_dim, 1024), nn.GELU(),
            nn.Dropout(0.30),
            nn.Linear(1024, 256), nn.GELU(),
            nn.Dropout(0.20),
            nn.Linear(256, num_classes),
        )

    def forward(self, clock, copy, trail, return_attention=False):
        f_clock = self.backbone(clock)
        f_copy = self.backbone(copy)
        f_trail = self.backbone(trail)

        original_stack = torch.stack([f_clock, f_copy, f_trail], dim=1)
        x = self.project(original_stack)
        x_with_modality = x + self.modality_embedding
        transformer_out = self.transformer(x_with_modality)
        fused, attention_weights = self.fusion(transformer_out, original_stack)
        logits = self.classifier(fused)

        if return_attention:
            return logits, attention_weights
        return logits


# ==========================================================
# LOAD TRAINED WEIGHTS (NO TRAINING)
# ==========================================================

model = MCITransformerConcatFusion(
    backbone_name=BACKBONE, num_classes=NUM_CLASSES, embed_dim=EMBED_DIM,
    num_heads=NUM_HEADS, num_layers=NUM_TRANSFORMER_LAYERS,
    ffn_dim=TRANSFORMER_FFN_DIM, transformer_dropout=TRANSFORMER_DROPOUT,
    attention_hidden=ATTENTION_HIDDEN, unfreeze_fraction=UNFREEZE_FRACTION,
).to(DEVICE)

assert os.path.exists(CHECKPOINT_PATH), f"Checkpoint not found at {CHECKPOINT_PATH}"
model.load_state_dict(torch.load(CHECKPOINT_PATH, map_location=DEVICE))
model.eval()
print(f"Loaded trained weights from {CHECKPOINT_PATH}")

# ==========================================================
# RE-DERIVE best_threshold ON VAL SET (same logic as training script)
# ==========================================================

val_probs, val_labels = [], []
with torch.no_grad():
    for batch in tqdm(val_loader, desc="Re-deriving threshold on val set"):
        clock = batch["clock"].to(DEVICE)
        copy = batch["copy"].to(DEVICE)
        trail = batch["trail"].to(DEVICE)
        labels = batch["label"].cpu().numpy()
        outputs = model(clock, copy, trail)
        probs = torch.softmax(outputs, dim=1)[:, 1]
        val_probs.extend(probs.cpu().numpy())
        val_labels.extend(labels)

best_threshold, best_f1 = 0.50, 0
for t in np.arange(0.20, 0.81, 0.01):
    preds = (np.array(val_probs) >= t).astype(int)
    f1 = f1_score(val_labels, preds)
    if f1 > best_f1:
        best_f1, best_threshold = f1, t

print(f"Best Threshold = {best_threshold:.2f} (Val F1 = {best_f1:.4f})")

# ==========================================================
# FULL MODEL TEST EVAL (all drawings) — for the comparison row
# ==========================================================

test_probs, test_preds, test_labels = [], [], []
with torch.no_grad():
    for batch in tqdm(test_loader, desc="Full model test eval"):
        clock = batch["clock"].to(DEVICE)
        copy = batch["copy"].to(DEVICE)
        trail = batch["trail"].to(DEVICE)
        labels = batch["label"].to(DEVICE)
        outputs = model(clock, copy, trail)
        probs = torch.softmax(outputs, dim=1)[:, 1]
        preds = (probs >= best_threshold).long()
        test_probs.extend(probs.cpu().numpy())
        test_preds.extend(preds.cpu().numpy())
        test_labels.extend(labels.cpu().numpy())

test_probs = np.array(test_probs)
test_preds = np.array(test_preds)
test_labels = np.array(test_labels)

accuracy = accuracy_score(test_labels, test_preds)
precision = precision_score(test_labels, test_preds)
recall = recall_score(test_labels, test_preds)
f1 = f1_score(test_labels, test_preds)
test_auc = roc_auc_score(test_labels, test_probs)
cm = confusion_matrix(test_labels, test_preds)
tn, fp, fn, tp = cm.ravel()
specificity = tn / (tn + fp)

print("\nFull model (all drawings) test results:")
print(f"  Accuracy={accuracy:.4f}  Precision={precision:.4f}  Recall={recall:.4f}  "
      f"Specificity={specificity:.4f}  F1={f1:.4f}  AUC={test_auc:.4f}")

# ==========================================================
# DRAWING-WISE (SINGLE-DRAWING) ABLATION
# ==========================================================

def make_blank_batch(batch_size, image_size, mean, std, device):
    blank = torch.full((batch_size, 3, image_size, image_size), 0.5, device=device)
    mean_t = torch.tensor(mean, device=device).view(1, 3, 1, 1)
    std_t = torch.tensor(std, device=device).view(1, 3, 1, 1)
    return (blank - mean_t) / std_t


def evaluate_single_drawing(model, loader, active_drawing, threshold,
                             image_size, mean, std, device):
    """Returns (metrics_dict, y_true, y_prob, y_pred) -- the raw arrays are
    needed by plot_confusion_matrices() / plot_roc_curves() below, not just
    the summary metrics."""
    assert active_drawing in ("clock", "copy", "trail")
    model.eval()
    y_true, y_prob, y_pred = [], [], []

    with torch.no_grad():
        for batch in tqdm(loader, leave=False, desc=f"Single-drawing: {active_drawing}"):
            real = {
                "clock": batch["clock"].to(device, non_blocking=True),
                "copy": batch["copy"].to(device, non_blocking=True),
                "trail": batch["trail"].to(device, non_blocking=True),
            }
            bsz = real["clock"].size(0)
            blank = make_blank_batch(bsz, image_size, mean, std, device)

            inputs = {}
            for name in ("clock", "copy", "trail"):
                inputs[name] = real[name] if name == active_drawing else blank

            labels = batch["label"].to(device)
            outputs = model(inputs["clock"], inputs["copy"], inputs["trail"])
            probs = torch.softmax(outputs, dim=1)[:, 1]
            preds = (probs >= threshold).long()

            y_true.extend(labels.cpu().numpy())
            y_prob.extend(probs.cpu().numpy())
            y_pred.extend(preds.cpu().numpy())

    y_true = np.array(y_true); y_prob = np.array(y_prob); y_pred = np.array(y_pred)
    cm_local = confusion_matrix(y_true, y_pred)
    tn_l, fp_l, fn_l, tp_l = cm_local.ravel()
    specificity_l = tn_l / (tn_l + fp_l) if (tn_l + fp_l) > 0 else float("nan")

    try:
        auc_val = roc_auc_score(y_true, y_prob)
    except Exception:
        auc_val = float("nan")

    metrics = {
        "Drawing": active_drawing.capitalize(),
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, zero_division=0),
        "Recall": recall_score(y_true, y_pred, zero_division=0),
        "Specificity": specificity_l,
        "F1": f1_score(y_true, y_pred, zero_division=0),
        "AUC": auc_val,
    }
    return metrics, y_true, y_prob, y_pred


results = []
# raw_by_condition holds (y_true, y_prob, y_pred) for every condition so we
# can build confusion matrices / ROC curves for all of them below.
raw_by_condition = {"All drawings (Proposed)": (test_labels, test_probs, test_preds)}

for drawing in ("clock", "copy", "trail"):
    res, y_true_d, y_prob_d, y_pred_d = evaluate_single_drawing(
        model=model, loader=test_loader, active_drawing=drawing,
        threshold=best_threshold, image_size=IMAGE_SIZE,
        mean=IMAGENET_MEAN, std=IMAGENET_STD, device=DEVICE,
    )
    results.append(res)
    raw_by_condition[res["Drawing"]] = (y_true_d, y_prob_d, y_pred_d)
    print(f"\n{drawing.upper()} only: {res}")

results.append({
    "Drawing": "All drawings (Proposed)",
    "Accuracy": accuracy, "Precision": precision, "Recall": recall,
    "Specificity": specificity, "F1": f1, "AUC": test_auc,
})

drawing_wise_df = pd.DataFrame(results)[
    ["Drawing", "Accuracy", "Precision", "Recall", "Specificity", "F1", "AUC"]
]

print("\n" + "=" * 70)
print("DRAWING-WISE (SINGLE-DRAWING) ABLATION RESULTS")
print("=" * 70)
print(drawing_wise_df.to_string(index=False))

drawing_wise_path = os.path.join(OUTPUT_DIR, "drawing_wise_ablation.csv")
drawing_wise_df.to_csv(drawing_wise_path, index=False)
print(f"\nSaved to: {drawing_wise_path}")

metric_cols = ["Accuracy", "Recall", "Specificity", "F1", "AUC"]
x = np.arange(len(drawing_wise_df))
width = 0.15

plt.figure(figsize=(10, 6))
for i, m in enumerate(metric_cols):
    plt.bar(x + i * width, drawing_wise_df[m], width, label=m)
plt.xticks(x + width * (len(metric_cols) - 1) / 2, drawing_wise_df["Drawing"], rotation=15)
plt.ylabel("Score")
plt.ylim(0, 1.05)
plt.title(f"Drawing-wise Ablation — {BACKBONE} + Transformer (Exp. B)")
plt.legend()
plt.grid(True, axis="y", alpha=0.3)
drawing_wise_bar_path = os.path.join(OUTPUT_DIR, "drawing_wise_ablation_bar.png")
plt.savefig(drawing_wise_bar_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"Bar chart saved to: {drawing_wise_bar_path}")

# ==========================================================
# CONFUSION MATRICES — one per condition (Clock, Copy, Trail, All)
# ==========================================================

def plot_confusion_matrices(raw_by_condition, class_names, output_dir):
    # Fixed, readable order: single drawings first, then the full model.
    order = ["Clock", "Copy", "Trail", "All drawings (Proposed)"]
    order = [c for c in order if c in raw_by_condition]  # keep only what we have

    n = len(order)
    ncols = 2
    nrows = int(np.ceil(n / ncols))

    fig, axes = plt.subplots(nrows, ncols, figsize=(5 * ncols, 4.5 * nrows))
    axes = np.array(axes).reshape(-1)

    cm_records = []
    for i, cond in enumerate(order):
        y_true_c, _, y_pred_c = raw_by_condition[cond]
        cm_c = confusion_matrix(y_true_c, y_pred_c, labels=[0, 1])
        tn_c, fp_c, fn_c, tp_c = cm_c.ravel()
        cm_records.append({"Condition": cond, "TN": int(tn_c), "FP": int(fp_c),
                            "FN": int(fn_c), "TP": int(tp_c)})

        ax = axes[i]
        im = ax.imshow(cm_c, cmap="Blues")
        ax.set_xticks([0, 1]); ax.set_xticklabels(class_names)
        ax.set_yticks([0, 1]); ax.set_yticklabels(class_names)
        ax.set_xlabel("Predicted"); ax.set_ylabel("True")
        ax.set_title(cond)
        for r in range(2):
            for c in range(2):
                ax.text(c, r, cm_c[r, c], ha="center", va="center",
                         color="white" if cm_c[r, c] > cm_c.max() / 2 else "black")

    # Hide any unused subplot slots (e.g. odd number of conditions)
    for j in range(len(order), len(axes)):
        axes[j].axis("off")

    fig.suptitle(f"Confusion Matrices by Drawing Condition — {BACKBONE} + Transformer", y=1.02)
    plt.tight_layout()
    cm_grid_path = os.path.join(output_dir, "confusion_matrices_all_drawings.png")
    plt.savefig(cm_grid_path, dpi=300, bbox_inches="tight")
    plt.show()

    cm_csv_path = os.path.join(output_dir, "confusion_matrices_all_drawings.csv")
    pd.DataFrame(cm_records).to_csv(cm_csv_path, index=False)

    print(f"\nConfusion matrix grid saved to: {cm_grid_path}")
    print(f"Confusion matrix counts saved to: {cm_csv_path}")
    return pd.DataFrame(cm_records)


cm_summary_df = plot_confusion_matrices(raw_by_condition, CLASS_NAMES, OUTPUT_DIR)
print("\nConfusion matrix counts (TN/FP/FN/TP) per condition:")
print(cm_summary_df.to_string(index=False))

# ==========================================================
# ROC CURVES — one combined overlay + per-condition CSV points
# ==========================================================

def plot_roc_curves(raw_by_condition, output_dir):
    order = ["Clock", "Copy", "Trail", "All drawings (Proposed)"]
    order = [c for c in order if c in raw_by_condition]

    plt.figure(figsize=(6.5, 6.5))
    roc_points = []

    for cond in order:
        y_true_c, y_prob_c, _ = raw_by_condition[cond]
        fpr, tpr, thresh = roc_curve(y_true_c, y_prob_c)
        auc_c = roc_auc_score(y_true_c, y_prob_c)
        plt.plot(fpr, tpr, label=f"{cond} (AUC = {auc_c:.3f})", linewidth=2)

        for f_val, t_val, th_val in zip(fpr, tpr, thresh):
            roc_points.append({"Condition": cond, "FPR": f_val, "TPR": t_val,
                                "Threshold": th_val})

    plt.plot([0, 1], [0, 1], "--", color="gray", label="Chance (AUC = 0.500)")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title(f"ROC Curves by Drawing Condition — {BACKBONE} + Transformer")
    plt.legend(loc="lower right")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()

    roc_plot_path = os.path.join(output_dir, "roc_curves_all_drawings.png")
    plt.savefig(roc_plot_path, dpi=300, bbox_inches="tight")
    plt.show()

    roc_csv_path = os.path.join(output_dir, "roc_curve_points.csv")
    pd.DataFrame(roc_points).to_csv(roc_csv_path, index=False)

    print(f"\nCombined ROC curve plot saved to: {roc_plot_path}")
    print(f"Per-point ROC data (fpr/tpr/threshold per condition) saved to: {roc_csv_path}")


plot_roc_curves(raw_by_condition, OUTPUT_DIR)

print("\n" + "=" * 70)
print("ALL DONE.")
print(f"  Drawing-wise metrics table : {drawing_wise_path}")
print(f"  Drawing-wise bar chart     : {drawing_wise_bar_path}")
print(f"  Confusion matrix grid      : {os.path.join(OUTPUT_DIR, 'confusion_matrices_all_drawings.png')}")
print(f"  Confusion matrix counts    : {os.path.join(OUTPUT_DIR, 'confusion_matrices_all_drawings.csv')}")
print(f"  ROC curve overlay          : {os.path.join(OUTPUT_DIR, 'roc_curves_all_drawings.png')}")
print(f"  ROC curve raw points       : {os.path.join(OUTPUT_DIR, 'roc_curve_points.csv')}")
print("=" * 70)
# ---- add at the very end of the script ----
lab = np.array(raw_by_condition["All drawings (Proposed)"][0])
df = pd.DataFrame({"patient_id": test_dataset.df["patient_id"].values, "label": lab})
for key, col in [("All drawings (Proposed)", "prob_all"), ("Clock", "prob_clock"),
                 ("Copy", "prob_copy"), ("Trail", "prob_trail")]:
    y_c, p_c, _ = raw_by_condition[key]
    assert (np.array(y_c) == lab).all()
    df[col] = np.array(p_c)
df["threshold"] = best_threshold
out = os.path.join(OUTPUT_DIR, "test_predictions_all_conditions.csv")
df.to_csv(out, index=False)
print("Saved:", out, "| n =", len(df), "| MCI =", int(df.label.sum()))
```

Statistical Test

```python
import os
import numpy as np
import pandas as pd
from scipy import stats
from sklearn.metrics import (
    roc_auc_score, accuracy_score, precision_score,
    recall_score, f1_score, confusion_matrix
)

CSV = "/content/drive/MyDrive/MCI_Results_efficientnet_b3_TransformerConcatFusion/test_predictions_all_conditions.csv"
OUT = os.path.dirname(CSV)

N_BOOT = 2000
SEED = 42
ALPHA = 0.05

# ---------------------------------------------------------
# Load data
# ---------------------------------------------------------
df = pd.read_csv(CSV)

y = df["label"].values
thr = float(df["threshold"].iloc[0])

CONDS = {
    "All drawings (Proposed)": "prob_all",
    "Clock only": "prob_clock",
    "Copy only": "prob_copy",
    "Trail only": "prob_trail"
}

# Check required columns
required = ["label", "threshold",
            "prob_all", "prob_clock",
            "prob_copy", "prob_trail"]

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(f"Missing columns: {missing}")

print("=" * 70)
print("STATISTICAL ANALYSIS")
print("=" * 70)
print(f"Patients: {len(df)}")
print(f"Threshold: {thr:.4f}")
print(f"Bootstrap samples requested: {N_BOOT}")
print("=" * 70)


# ---------------------------------------------------------
# Metrics
# ---------------------------------------------------------
def metrics(y, p, thr):

    pred = (p >= thr).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y, pred, labels=[0, 1]
    ).ravel()

    return {
        "Accuracy": accuracy_score(y, pred),
        "Precision": precision_score(
            y, pred, zero_division=0
        ),
        "Recall": recall_score(
            y, pred, zero_division=0
        ),
        "Specificity": (
            tn / (tn + fp)
            if (tn + fp) > 0 else np.nan
        ),
        "F1": f1_score(
            y, pred, zero_division=0
        ),
        "AUC": roc_auc_score(y, p)
    }


# ---------------------------------------------------------
# Holm correction
# ---------------------------------------------------------
def holm(pvals):

    p = np.asarray(pvals, dtype=float)

    order = np.argsort(p)
    m = len(p)

    adjusted = np.empty(m)
    running = 0.0

    for rank, idx in enumerate(order):

        value = (m - rank) * p[idx]

        running = max(running, value)

        adjusted[idx] = min(1.0, running)

    return adjusted


# ---------------------------------------------------------
# DeLong implementation
# ---------------------------------------------------------
def _midrank(x):

    j = np.argsort(x)
    z = x[j]

    n = len(x)
    t = np.zeros(n)

    i = 0

    while i < n:

        k = i

        while k < n and z[k] == z[i]:
            k += 1

        t[i:k] = 0.5 * (i + k - 1) + 1

        i = k

    out = np.empty(n)
    out[j] = t

    return out


def _delong_parts(y, p):

    pos = p[y == 1]
    neg = p[y == 0]

    m = len(pos)
    n = len(neg)

    tz = _midrank(
        np.concatenate([pos, neg])
    )

    tx = _midrank(pos)
    ty = _midrank(neg)

    auc = (
        tz[:m].sum() / (m * n)
        - (m + 1) / (2 * n)
    )

    vx = (tz[:m] - tx) / n
    vy = 1 - (tz[m:] - ty) / m

    return auc, vx, vy


def delong_paired(y, p1, p2):

    auc1, x1, y1 = _delong_parts(y, p1)
    auc2, x2, y2 = _delong_parts(y, p2)

    m = len(x1)
    n = len(y1)

    Sx = np.cov(
        np.vstack([x1, x2])
    ) / m

    Sy = np.cov(
        np.vstack([y1, y2])
    ) / n

    S = Sx + Sy

    variance = (
        S[0, 0]
        + S[1, 1]
        - 2 * S[0, 1]
    )

    if variance <= 0:
        return auc1 - auc2, 1.0

    z = (auc1 - auc2) / np.sqrt(variance)

    p_value = 2 * stats.norm.sf(abs(z))

    return auc1 - auc2, p_value


# ---------------------------------------------------------
# Exact McNemar
# ---------------------------------------------------------
def mcnemar_exact(y, p1, p2, thr):

    pred1 = (p1 >= thr).astype(int)
    pred2 = (p2 >= thr).astype(int)

    correct1 = pred1 == y
    correct2 = pred2 == y

    b = int(
        (correct1 & ~correct2).sum()
    )

    c = int(
        (~correct1 & correct2).sum()
    )

    if b + c == 0:
        p_value = 1.0
    else:
        p_value = stats.binomtest(
            b,
            b + c,
            0.5
        ).pvalue

    return b, c, p_value


# ---------------------------------------------------------
# Patient-level bootstrap indices
# ---------------------------------------------------------
rng = np.random.RandomState(SEED)

n = len(y)

boot_idx = []

while len(boot_idx) < N_BOOT:

    idx = rng.randint(0, n, n)

    # AUC requires both classes
    if 0 < y[idx].sum() < n:
        boot_idx.append(idx)

print(
    f"Valid bootstrap samples: {len(boot_idx)}"
)


# =========================================================
# 1. Bootstrap 95% CIs
# =========================================================

rows = []

for name, col in CONDS.items():

    p = df[col].values

    point = metrics(y, p, thr)

    bootstrap_results = []

    for idx in boot_idx:

        bootstrap_results.append(
            metrics(
                y[idx],
                p[idx],
                thr
            )
        )

    bs = pd.DataFrame(
        bootstrap_results
    )

    row = {
        "Condition": name
    }

    for metric_name in point:

        lo, hi = np.nanpercentile(
            bs[metric_name],
            [2.5, 97.5]
        )

        row[metric_name] = (
            f"{point[metric_name]:.3f} "
            f"[{lo:.3f}, {hi:.3f}]"
        )

    rows.append(row)


ci_table = pd.DataFrame(rows)

ci_path = os.path.join(
    OUT,
    "stat_ci_table.csv"
)

ci_table.to_csv(
    ci_path,
    index=False
)

print("\n")
print("=" * 70)
print("95% BOOTSTRAP CONFIDENCE INTERVALS")
print("=" * 70)

print(
    ci_table.to_string(index=False)
)

print(f"\nSaved: {ci_path}")


# =========================================================
# 2. Paired comparisons
# =========================================================

p_full = df["prob_all"].values

pairwise = []

for name, col in CONDS.items():

    if col == "prob_all":
        continue

    p_single = df[col].values

    # -----------------------------------------------------
    # DeLong test
    # -----------------------------------------------------
    d_auc, p_delong = delong_paired(
        y,
        p_full,
        p_single
    )

    # -----------------------------------------------------
    # McNemar test
    # -----------------------------------------------------
    b, c, p_mcnemar = mcnemar_exact(
        y,
        p_full,
        p_single,
        thr
    )

    # -----------------------------------------------------
    # Paired bootstrap CI for AUC difference
    # -----------------------------------------------------
    auc_diffs = []

    for idx in boot_idx:

        auc_full = roc_auc_score(
            y[idx],
            p_full[idx]
        )

        auc_single = roc_auc_score(
            y[idx],
            p_single[idx]
        )

        auc_diffs.append(
            auc_full - auc_single
        )

    auc_diffs = np.asarray(auc_diffs)

    lo, hi = np.percentile(
        auc_diffs,
        [2.5, 97.5]
    )

    pairwise.append({

        "Comparison":
            f"All vs {name}",

        "AUC difference":
            d_auc,

        "AUC difference 95% CI":
            f"[{lo:.3f}, {hi:.3f}]",

        "DeLong p":
            p_delong,

        "McNemar b":
            b,

        "McNemar c":
            c,

        "McNemar exact p":
            p_mcnemar
    })


pairwise = pd.DataFrame(pairwise)


# ---------------------------------------------------------
# Holm correction
# ---------------------------------------------------------

pairwise["DeLong p (Holm)"] = holm(
    pairwise["DeLong p"].values
)

pairwise["McNemar p (Holm)"] = holm(
    pairwise["McNemar exact p"].values
)


# ---------------------------------------------------------
# Save
# ---------------------------------------------------------

pairwise_path = os.path.join(
    OUT,
    "stat_pairwise_tests.csv"
)

pairwise.to_csv(
    pairwise_path,
    index=False
)


print("\n")
print("=" * 70)
print("PAIRED STATISTICAL COMPARISONS")
print("=" * 70)

print(
    pairwise.round(4).to_string(
        index=False
    )
)

print(f"\nSaved: {pairwise_path}")


# =========================================================
# 3. Significance summary
# =========================================================

print("\n")
print("=" * 70)
print("SIGNIFICANCE AFTER HOLM CORRECTION")
print("=" * 70)

for _, r in pairwise.iterrows():

    print(
        f"\n{r['Comparison']}"
    )

    print(
        f"  DeLong: "
        f"p = {r['DeLong p (Holm)']:.4f} "
        f"-> "
        f"{'SIGNIFICANT' if r['DeLong p (Holm)'] < ALPHA else 'not significant'}"
    )

    print(
        f"  McNemar: "
        f"p = {r['McNemar p (Holm)']:.4f} "
        f"-> "
        f"{'SIGNIFICANT' if r['McNemar p (Holm)'] < ALPHA else 'not significant'}"
    )
```
